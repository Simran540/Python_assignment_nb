```python
# 1. Write a program to find greatest of four numbers entered by the user.
a = int(input("Enter a: "))
b = int(input("Enter b: "))
c = int(input("Enter c: "))
d = int(input("Enter d: "))

print("Greatest:", max(a, b, c, d))
```

```python
# 2. Write a program to find out whether a student is pass or fail.
m1 = int(input("Enter marks 1: "))
m2 = int(input("Enter marks 2: "))
m3 = int(input("Enter marks 3: "))

total = (m1 + m2 + m3) / 3

if total >= 40 and m1 >= 33 and m2 >= 33 and m3 >= 33:
    print("Pass")
else:
    print("Fail")
```

```python
# 3. A spam comment is defined as a text containing following keywords:
comment = input("Enter comment: ")

if ("make a lot of money" in comment or
    "buy now" in comment or
    "subscribe this" in comment or
    "click this" in comment):
    print("Spam detected")
else:
    print("Not spam")
```

```python
# 4. Write a program to find whether a given username contains less than 10 characters or not.
username = input("Enter username: ")

if len(username) < 10:
    print("Username contains less than 10 characters")
else:
    print("Username contains 10 or more characters")
```

```python
# 1. Write a program to print multiplication table of a given number using for loop. ex: 1 X 1 =1

n = int(input("Enter number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n*i)
```

```python
# 2. Write a program to greet all the person names stored in a list l1 and which starts with S
#Example list:
#  l1 = ["Harry", "Soham", "Sachin", "Rahul"]

l1 = ["Harry", "Soham", "Sachin", "Rahul"]

for name in l1:
    if name.startswith("S"):
        print("Hello", name)
```

```python
# 3. Write a program to calculate the factorial of a given number using for loop.
n = int(input("Enter number: "))
fact = 1

for i in range(1, n+1):
    fact *= i

print("Factorial:", fact)
```

```python
# 4. Write a program to print the following star pattern (for n = 3)

n = 3
for i in range(1, n + 1):
    print("* " * i)
```

```python
# 5. Write a program to print the following star pattern (for n = 3)
n = 3
for i in range(1, n+1):
    print(" " * (n-i) + "* " * i)


```
